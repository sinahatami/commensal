```python
import pandas as pd

file_path = '../video_2.csv'
delimiter = ',' 

annotations_df = pd.read_csv(file_path, delimiter=delimiter)

print(annotations_df.head())
```

```python
label_counts = annotations_df['Tier'].value_counts()

print("Frequency of Each Label:")
print(label_counts)
```

```python
overlapping_annotations = annotations_df.groupby(['Start Time', 'End Time']).size().reset_index(name='Count')

overlapping_annotations = overlapping_annotations[overlapping_annotations['Count'] > 1]

print("Overlapping Annotations:")
print(overlapping_annotations)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
label_counts.plot(kind='bar')
plt.title('Frequency of Each Label')
plt.xlabel('Label')
plt.ylabel('Frequency')
plt.show()
```
