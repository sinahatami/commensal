```python
import pympi
import csv
import os

def convert_eaf_to_csv(eaf_file_path, output_directory):
    # Ensure output directory exists
    os.makedirs(output_directory, exist_ok=True)

    # Extract the filename and create the CSV file path
    base_filename = os.path.basename(eaf_file_path).replace('.eaf', '.csv')
    csv_file_path = os.path.join(output_directory, base_filename)

    # Load the EAF file
    elan_file = pympi.Elan.Eaf(eaf_file_path)

    # Open the CSV file for writing
    with open(csv_file_path, mode='w', newline='', encoding='utf-8') as csv_file:
        csv_writer = csv.writer(csv_file)
        
        # Write the header
        csv_writer.writerow(['Tier', 'Start Time', 'End Time', 'Annotation'])
        
        # Iterate over all tiers in the EAF file
        for tier_name in elan_file.get_tier_names():
            for annotation in elan_file.get_annotation_data_for_tier(tier_name):
                start_time = annotation[0]
                end_time = annotation[1]
                annotation_value = annotation[2]
                # Write each annotation to the CSV file
                csv_writer.writerow([tier_name, start_time, end_time, annotation_value])
    
    print(f"Conversion complete. The data has been saved to {csv_file_path}")

```

```python
# Directory to save output CSV files
output_directory = './final_dataset/csv'

# List of EAF files to convert
eaf_files = ['./final_dataset/eaf/video_20.eaf',]

# Convert each EAF file to a CSV file
for eaf_file in eaf_files:
    convert_eaf_to_csv(eaf_file, output_directory)
```

```python
import pandas as pd
import os

def update_annotation(row):
    tier = row['Tier']
    annotation = row['Annotation']
    #print(annotation)
    if annotation == "Yes":
        return tier
    elif annotation == "Fork_towards":
        return "Left_Fork_towards" if "left" in tier.lower() else "Right_Fork_towards"
    elif annotation == "Look_at_plate":
        return "Left_Look_at_plate" if "left" in tier.lower() else "Right_Look_at_plate"
    elif annotation == "Look_towards":
        return "Left_Look_towards" if "left" in tier.lower() else "Right_Look_towards"
    else:
        return annotation

def process_csv_file(input_file, output_directory):
    df = pd.read_csv(input_file)
    df['Annotation'] = df.apply(update_annotation, axis=1)
    df.drop('Tier', axis=1, inplace=True)

    # Ensure output directory exists
    os.makedirs(output_directory, exist_ok=True)
    base_filename = os.path.basename(input_file)
    output_file = os.path.join(output_directory, base_filename)

    df.to_csv(output_file, index=False)
    print(f"File {input_file} has been processed and saved as {output_file}")

# Directory to save output CSV files
output_directory = './final_dataset/csv'

# List of CSV files to process
csv_files = ['./final_dataset/csv/video_20.csv']

# Process each CSV file
for csv_file in csv_files:
    process_csv_file(csv_file, output_directory)
```
